```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/sample_data/california_housing_train.csv')
df.head()
```

```python
df['population'].fillna(value=df['population'].mean(), inplace=True)
```

```python
df['households'].fillna(value="DECK", inplace=True)
```

```python
df['median_house_value'] = df['median_house_value'].fillna(df['median_house_value'].mode()[0])
```

```python
df.to_csv('/content/sample_data/california_housing_train.csv', index=False)
```
