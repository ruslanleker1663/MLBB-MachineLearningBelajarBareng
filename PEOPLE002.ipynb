```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

data load


```python
titanic_dataset_fixed_df = pd.read_csv('/content/sensus_penduduk_kalipare.csv')
df = titanic_dataset_fixed_df
df.head()
```

data slicing


```python
df = df[['Jumlah_Dusun', 'Jumlah_RW', 'Jumlah_RT', 'Estimasi_Penduduk_Laki_Laki', 'Estimasi_Penduduk_Perempuan']]
df.head()
```

encoding datas


```python
le = LabelEncoder()
df['Jumlah_RW'] = le.fit_transform(df['Jumlah_RW'])
df['Jumlah_RT'] = le.fit_transform(df['Jumlah_RT'])
```

result verifying


```python
df.head()
```

standard

```python
std = StandardScaler()
df['Estimasi_Penduduk_Perempuan'] = std.fit_transform(df[['Estimasi_Penduduk_Perempuan']])
```

```python
df.head()
```

```python
std = StandardScaler()
df['Estimasi_Penduduk_Laki_Laki'] = std.fit_transform(df[['Estimasi_Penduduk_Laki_Laki']])
```
