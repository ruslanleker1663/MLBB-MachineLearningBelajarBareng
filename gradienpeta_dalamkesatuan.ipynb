```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Define the coordinates relative to Gunung Arjuno (0,0)
x_points = np.array([-0.10925, 0, 0.15084])
y_points = np.array([-0.38863, 0, 0.39139])
labels = ['Jambuwer Village', 'Arjuno Mountain (0,0)', 'Ketintang Hometown']

# 2. Generate data points for the rounded linear equation: y = 3x - 0.1
x_line = np.linspace(-0.15, 0.20, 100)
y_line = 3 * x_line - 0.1

# 3. Create the plot
plt.figure(figsize=(8, 6))

# Plot the linear model line
plt.plot(x_line, y_line, color='crimson', linestyle='--', linewidth=2, label='Linear Model: $y = 3x - 0.1$')

# Scatter the specific geographic locations
plt.scatter(x_points, y_points, color='navy', s=80, zorder=5, label='Locations')

# Add text labels next to each point
for i, txt in enumerate(labels):
    plt.annotate(txt, (x_points[i], y_points[i]), textcoords="offset points", xytext=(0, 10), ha='center', fontsize=9, weight='bold')

# 4. Chart styling and formatting
plt.axhline(0, color='black', linewidth=0.8, linestyle='-') # X-axis baseline
plt.axvline(0, color='black', linewidth=0.8, linestyle='-') # Y-axis baseline
plt.grid(color='lightgray', linestyle=':', linewidth=0.5)

plt.title('Linear Alignment Relative to Mount Arjuno', fontsize=12, pad=15, weight='bold')
plt.xlabel('Longitude Offset (Degrees from Center)', fontsize=10)
plt.ylabel('Latitude Offset (Degrees from Center)', fontsize=10)
plt.legend(loc='upper left')

# Adjust layout and display the visualization
plt.tight_layout()
plt.show()
```
