# Titanic Dataset - Exploratory Data Analysis (EDA)

Notebook ini berisi langkah-langkah lengkap Analisis Data Eksploratif (EDA) untuk dataset Titanic.

## Step 1 - Loading Data & Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
# Load dataset Titanic
df = pd.read_csv('Titanic-Dataset.csv')
```

## Step 3 - Visual Inspection

### Data Distribution Inspection (Histogram)

```python
num_cols = ['Age', 'Fare', 'SibSp', 'Parch']

for col in num_cols:
    if col in df.columns:
        plt.figure(figsize=(7, 4))
        plt.hist(df[col].dropna(), bins=30, color='skyblue', edgecolor='black')
        plt.title(f'Distribusi {col}')
        plt.xlabel(col)
        plt.ylabel('Frequency')
        plt.tight_layout()
        plt.show()
```

### Outlier Inspection (Boxplot)

```python
for col in ['Age', 'Fare']:
    if col in df.columns:
        plt.figure(figsize=(6, 4))
        plt.boxplot(df[col].dropna(), orientation='vertical')
        plt.title(f'Boxplot {col}')
        plt.ylabel(col)
        plt.tight_layout()
        plt.show()
```

### Composition of Qualitative Variables (Bar Chart)

```python
cat_cols = ['Survived', 'Pclass', 'Sex', 'Embarked']

fig, axes = plt.subplots(2, 2, figsize=(11, 8))
axes = axes.flatten()

for i, col in enumerate(cat_cols):
    if col in df.columns:
        df[col].value_counts(dropna=False).plot(kind='bar', ax=axes[i], color='teal', edgecolor='black')
        axes[i].set_title(f'Distribusi {col}')
        axes[i].set_xlabel(col)
        axes[i].set_ylabel('Count')
        axes[i].tick_params(axis='x', rotation=0)

plt.tight_layout()
plt.show()
```
