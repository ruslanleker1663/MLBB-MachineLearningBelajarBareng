```python
import pandas as pd
import numpy as np
```

Data Load

```python
df = pd.read_csv('/content/sensus_penduduk_kalipare.csv')

df.head()
```

Imputation Datas

```python
df['Jumlah_RW'] = df['Jumlah_RW'].fillna(value=df['Jumlah_RW'].mean())
```

```python
df['Jumlah_RT'] = df['Jumlah_RT'].fillna(value="DECK")
```

```python
df['Total_Penduduk'] = df['Total_Penduduk'].fillna(df['Total_Penduduk'].mode()[0])
```

Result Validate

```python
df.info()
```

```python
df.to_csv('/content/sensus_penduduk_kalipare.csv', index=False)
```
